```python
import matplotlib.pyplot as plt 
import cv2
```

```python
img_path = '../../assets/dataset/test/clsdata(1031)/cls/images/val2019/fa164d26f664703f96ed07dda2a05b48_d3d3LnByZWYuZnVrdWkuanAJMjIwLjIxMy4xODguMw==.xls-1-0.png'
img = cv2.imread(img_path)
img
```

```python
%matplotlib inline
plt.figure(figsize=(10, 8))
cv2.rectangle(
    img,
    (302,   4), 
    (469,  34),
    (0, 255, 0), 
    thickness=2
)
plt.imshow(img)
plt.show()
```

```python
cropped_img = img[4:34, 302:469] 
plt.imshow(cropped_img)
plt.show()
```

```python
import pytesseract
```

```python
gray = cv2.cvtColor(cropped_img, cv2.COLOR_BGR2GRAY)
text = pytesseract.image_to_string(gray, lang="jpn")
print("Nội dung OCR:", text)
```
